```
def operacao (valor1, valor2, conta):
  if(conta == 1):
    return valor1 + valor2;
  if(conta == 2):
    return valor1 - valor2;
  if(conta == 3):
    return valor1 * valor2;
  if(conta == 4):
    return valor1 / valor2;

  else:
    return 0

resultado = operacao(8,2,4)
print(resultado)
```
